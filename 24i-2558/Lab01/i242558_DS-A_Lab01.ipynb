```python
import pandas as pd
import numpy as np

df = pd.read_csv('adult.csv')
df.info()
df.head()
```

```python
#Task 1
df.isnull().sum()   

df.replace('?', np.nan, inplace=True)

df.isnull().sum()   
```

```python
#Task 2
missing_count = df.isnull().sum()
missing_pct = df.isnull().mean() * 100

missing_summary = pd.DataFrame({'count': missing_count, 'percent': missing_pct})
missing_summary = missing_summary[missing_summary['count'] > 0].sort_values('count', ascending=False)
missing_summary
```

```python
# Task 3
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])
```

## Data Profiling Summary

**Dataset overview**
32,561 rows and 15 columns from the UCI Adult Income dataset

**Data quality issues found**
- Missing values hidden as literal "?" strings (not real NaN) in three columns:
  occupation , workclass, and native.country 
- 24 exact duplicate rows; 25 rows duplicate on all features except income
- No whitespace inconsistencies found in education, marital.status, or
  native.country in this version of the file.

**Key observations**
- Income classes are imbalanced: 75.9% earn <=50K vs 24.1% earning >50K
- Sex distribution is skewed: 66.9% Male vs 33.1% Female.
- Most common occupation is Prof-specialty (4,140), narrowly ahead of
  Craft-repair (4,099) and Exec-managerial (4,066).
- education and education.num are perfectly consistent, no mismatches.

**Cleaning decisions made**
- Replaced "?" with NaN to expose true missingness.
- Imputed workclass and occupation with "Unknown" rather than dropping,
  to preserve ~5.6% of rows for other features.
- Imputed native.country with its mode (United-States) given its low
  missing rate (1.79%) and heavy skew toward one category.
