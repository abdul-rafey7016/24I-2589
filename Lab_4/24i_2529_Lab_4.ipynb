```python
#Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#load dataset
df = pd.read_csv('sample_superstore.csv', encoding='latin1')
df
```

```python
#Task1
plt.figure(figsize=(10, 6))
sns.histplot(df['Sales'], kde=True)
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Distribution of Sales')
plt.show()

skewness = df['Sales'].skew()
mean = df['Sales'].mean()
median = df['Sales'].median()

print("Skewness:", skewness)
print("Mean:", mean)
print("Median:", median)

#The sales distribution is right-skewed if the skewness value is positive, with the mean greater than the median.
#This indicates that a few high-sales values pull the mean upward, so the median may be a better measure of the typical sale.
```

```python
#task2
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='Category', y='Profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.title('Profit Distribution by Category')
plt.show()

plt.figure(figsize=(10, 6))
sns.stripplot(data=df, x='Category', y='Profit', jitter=True)
plt.xlabel('Category')
plt.ylabel('Profit')
plt.title('Profit Distribution by Category - Strip Plot')
plt.show()

#The violin plot better reveals a category with a bimodal or unusually spread-out profit distribution because it shows the overall density and shape of the distribution.
#The strip plot shows individual observations clearly, but overlapping points can make the overall distribution shape harder to identify.
```

```python
#Task3
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit',
    hue='Category',
    size='Sales',
    sizes=(20, 300),
    alpha=0.6
)

plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit by Category and Sales')
plt.show()
#The plot shows how the relationship between discount and profit differs across categories.
#The size encoding also shows that orders with larger sales can have substantially different profit levels, making patterns that would not be visible in a simple discount-versus-profit scatter plot easier to identify.
```

```python
#Task4
numeric_df = df.select_dtypes(include='number')

pearson_corr = numeric_df.corr(method='pearson')
spearman_corr = numeric_df.corr(method='spearman')

print("Pearson Correlation Matrix:")
print(pearson_corr)

print("\nSpearman Correlation Matrix:")
print(spearman_corr)

plt.figure(figsize=(10, 6))
sns.heatmap(pearson_corr, annot=True, cmap='Blues', fmt='.2f')
plt.title('Pearson Correlation Matrix')
plt.show()

plt.figure(figsize=(10, 6))
sns.heatmap(spearman_corr, annot=True, cmap='Blues', fmt='.2f')
plt.title('Spearman Correlation Matrix')
plt.show()

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5)
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit')
plt.show()

#Pearson and Spearman correlations differ noticeably for profit and discount.
#The scatter plot suggests that the difference is caused by outliers, because Pearson measures linear correlation while
#Spearman measures the strength of a monotonic relationship. Correlation does not imply causation; two variables may be related because of other factors or by coincidence.
```

```python
#Task 5
sales_pivot = pd.pivot_table(
    df,
    index='Region',
    columns='Category',
    values='Sales',
    aggfunc='sum',
    margins=True
)

print(sales_pivot)

plt.figure(figsize=(10, 6))
sns.heatmap(
    sales_pivot,
    annot=True,
    fmt='.0f',
    cmap='Blues'
)

plt.xlabel('Category')
plt.ylabel('Region')
plt.title('Total Sales by Region and Category')
plt.show()

sales_only = sales_pivot.drop(index='All', columns='All')
max_region, max_category = sales_only.stack().idxmax()
max_sales = sales_only.stack().max()

print("\nHighest Sales Combination:")
print("Region:", max_region)
print("Category:", max_category)
print("Total Sales:", max_sales)

#The region and category combination contributes the most to overall revenue, with total sales of value.
#The heatmap makes this combination easy to identify because it has the highest value among the region–category cells.
```

```python
#Task6
df['Order Date'] = pd.to_datetime(df['Order Date'])

df = df.set_index('Order Date')

monthly_sales = df['Sales'].resample('ME').sum()
rolling_sales = monthly_sales.rolling(window=3).mean()

plt.figure(figsize=(12, 6))
plt.plot(monthly_sales, label='Monthly Sales')
plt.plot(rolling_sales, label='3-Month Rolling Average')

plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Monthly Sales with 3-Month Rolling Average')
plt.legend()
plt.show()
#The 3-month rolling average smooths out short-term fluctuations in monthly sales and makes the overall trend easier to see.
#It reduces month-to-month noise that is visible in the raw monthly sales line.
```

```python
#task7
df['Profitable'] = df['Profit'] > 0

g = sns.FacetGrid(
    df,
    col='Region',
    hue='Profitable',
    col_wrap=2,
    height=5
)

g.map_dataframe(
    sns.scatterplot,
    x='Discount',
    y='Profit',
    alpha=0.5
)

g.add_legend()
g.set_axis_labels('Discount', 'Profit')
g.set_titles('{col_name}')
plt.show()

#The region shows a visibly different discount–profit relationship compared with the other regions.
```

```python
#task 8
best = df.loc[df['Profit'].idxmax()].iloc[0]
worst = df.loc[df['Profit'].idxmin()].iloc[0]

plt.figure(figsize=(12, 7))

plt.scatter(
    df['Discount'],
    df['Profit'],
    alpha=0.2
)

plt.scatter(
    best['Discount'],
    best['Profit'],
    s=100,
    marker='o'
)

plt.annotate(
    f"Highest Profit: {best['Profit']:.0f}",
    xy=(best['Discount'], best['Profit']),
    xytext=(best['Discount'] + 0.05, best['Profit'] + 100),
    arrowprops=dict(arrowstyle='->')
)

plt.scatter(
    worst['Discount'],
    worst['Profit'],
    s=100,
    marker='o'
)

plt.annotate(
    f"Largest Loss: {worst['Profit']:.0f}",
    xy=(worst['Discount'], worst['Profit']),
    xytext=(worst['Discount'] + 0.05, worst['Profit'] - 100),
    arrowprops=dict(arrowstyle='->')
)

plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit with Extreme Orders Annotated')
plt.show()
#Transparency (alpha) was used because the large number of order-level points causes overplotting.
# Lowering the opacity makes overlapping points visible while preserving the individual observations.
```
