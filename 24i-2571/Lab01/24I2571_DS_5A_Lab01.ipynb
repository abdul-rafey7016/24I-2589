**Justification**:



I replaced the missing values in occupation with Unknown because we cannot determine the actual occupation, and using the mode could assign an incorrect occupation.

I also replaced the missing values in workclass with Unknown for the same reason. This keeps all the records in the dataset without making assumptions about the missing values.

There were no missing values in the numerical columns, so no numerical imputation was required.

```python
#TASK 9

import pandas as pd
import sqlite3


df_original = pd.read_csv('/content/drive/MyDrive/adult.csv')

print("Original CSV shape:", df_original.shape)


conn = sqlite3.connect("adult_income.db")


df_original.to_sql(
    "adult_income",
    conn,
    if_exists="replace",
    index=False
)


query = """
SELECT *
FROM adult_income
WHERE age > 30
"""

df_sql = pd.read_sql_query(query, conn)


df_pandas = df_original[df_original["age"] > 30]


print("\n--- Shape Verification ---")
print("Original CSV shape:", df_original.shape)
print("SQL result shape:", df_sql.shape)
print("Pandas filtered shape:", df_pandas.shape)

print("\n--- Condition Verification ---")
print("All extracted ages > 30:", (df_sql["age"] > 30).all())
print("Minimum extracted age:", df_sql["age"].min())


df_sql_sorted = df_sql.sort_values(
    by=list(df_sql.columns)
).reset_index(drop=True)

df_pandas_sorted = df_pandas.sort_values(
    by=list(df_pandas.columns)
).reset_index(drop=True)

print("\n--- Data Verification ---")
print("SQL and pandas results are identical:",
      df_sql_sorted.equals(df_pandas_sorted))


print("\n--- Extracted Data ---")
print(df_sql.head())


conn.close()
```
