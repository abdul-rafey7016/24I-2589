Task 1

```python
import pandas as pd
import numpy as np

df = pd.read_csv('adult.csv')

print("initial information")
df.info()

print("\nfirst five rows")
print(df.head())

print("\nnull count before fix")
print(df.isnull().sum())


df.replace('?', np.nan, inplace=True)

print("\nnull count after fix")
print(df.isnull().sum())
```

Task 2

```python
missingvals = df.isnull().sum()

missingpercent = df.isnull().mean() * 100

report = pd.DataFrame({
    'Missing Count': missingvals,
    'Missing Percentage (%)': missingpercent
})

# sort the values
sortedreport = report.sort_values(by='Missing Count', ascending=False)

print("Missing values and the percentage sorted")
print(sortedreport)

print("\ncolumns with most missing values")
print("occupation, workclass, native.country (in that order)")
```

Task 3

```python
# for categorical, fill with unkown
catcols = ['workclass', 'occupation', 'native.country']
for col in catcols:
    if col in df.columns:
        df[col].fillna('Unknown', inplace=True)

# fill numerical with medians
numericcols = df.select_dtypes(include=['int64', 'float64']).columns
for col in numericcols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)

print("Verify if any missing left:")
print(df.isnull().sum().sum(), "total missing values remaining.")
```

Task 4

```python
duplicates = df.duplicated().sum()
print(f"Exact duplicate rows: {duplicates}")

colsnoincome = [col for col in df.columns if col != 'income'] #creating a list of all columns to ignore income (the target col)

# find duplicates now
duplicateswithoutincome = df.duplicated(subset=colsnoincome).sum()
print(f"Duplicates ignoring the 'income' column: {duplicateswithoutincome}")

#drop only exact duplicates
df.drop_duplicates(inplace=True)
print(f"Rows remaining after dropping exact duplicates: {len(df)}")
```

Task 5

```python
cols_to_clean = ['education', 'marital.status', 'native.country']

print("unique values before cleaning")
for col in cols_to_clean:
    print(f"\n{col} unique values:")
    print(df[col].unique())


for col in cols_to_clean:
    df[col] = df[col].astype(str).str.strip() # astype to ensure that data type is of string

#verification
print("\nafter cleaning")
for col in cols_to_clean:
    print(f"\n{col} unique values:")
    print(df[col].unique())
```

Task 6

```python
print("summary of numerical columns")
print(df.describe())

print("\ncategorical summary")
print(df.describe(include='object'))

age_mean = df['age'].mean()
age_median = df['age'].median()

hours_mean = df['hours.per.week'].mean()
hours_median = df['hours.per.week'].median()

print(f"Age Mean: {age_mean:.2f}, Age Median: {age_median:.2f}")
print(f"Hours Mean: {hours_mean:.2f}, Hours Median: {hours_median:.2f}")
```

Task 7

```python
print("Most Common Occupation")
occupation_counts = df['occupation'].value_counts()
print(occupation_counts.head(1))

print("\nGender Dist.")
genderdist = df['sex'].value_counts(normalize=True) * 100
print(genderdist)

# c) Percentage distribution of the income label
print("\nIncome Dist.")
incomedist = df['income'].value_counts(normalize=True) * 100
print(incomedist)
```

```python
edumap = df.groupby('education')['education.num'].unique()

print("Mapping")
print(edumap)


inconsistencies = edumap[edumap.apply(len) > 1]

print("\nCheck inconsistency")
if len(inconsistencies) == 0:
    print("No inconsistencies in data.")
else:
    print("Inconsistencies found:")
    print(inconsistencies)
```

Task 9

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('adult_income_db.sqlite')

raw_df = pd.read_csv('adult.csv')
raw_df.to_sql('adult_income', conn, if_exists='replace', index=False)


sql_query = "SELECT * FROM adult_income WHERE age > 30"

sql_df = pd.read_sql_query(sql_query, conn)

print("Shape compared.")
print(f"Original CSV Shape (All Data): {raw_df.shape}")
print(f"SQL Extracted DataFrame Shape (Age > 30): {sql_df.shape}")

conn.close()
```

The dataset has 15 columns and 32561 rows prior to cleaning and dropping etc. It is representing data of different adults and their incomes.

Issues with data:
Missing data was denoted by '?' instead of actual NaN values. In reality, there are 1800+ missing values. It also had some duplicates across.

1 to 1 mapping of education was verified.

Cleaning decisions:
Missing values of categorical type were replaced with "unkown" to retain the rows, while numerical ones were replaced with median.
Duplicates that were exact were dropped, but those differing only on income were kept.
Inconsitent data was cleaned up using str.strip, but I didn't observe any significant change.
