```python
import numpy as np
import pandas as pd

#task1
df=pd.read_csv("adult.csv")
print("info")
df.info()
print("null is ")
print(df.isnull().sum())
df.replace("?",np.nan, inplace=True)

#task2
print("descending order ")
print(df.isnull().sum().sort_values(ascending=False))


#task3
print("For workclass")
print(df['workclass'].replace("?", "unknown",inplace=False).fillna("unknown"))
print("for occupation replacing")
print(df['occupation'].replace("?", "unknown",inplace=False).fillna("unknown"))
#replacing with unknown because we dont know for sure that mode will be the correct answer
print("no numeric column having ?")

#task4
print("exact duplicated rows are ")
print(df.duplicated().sum())

f=df.columns.drop("age")
f_duplicates=df.duplicated(subset=f).sum()
print("duplicates without any label ", f_duplicates)

#task5
print("unique values in education")
print(df['education'].str.strip().unique())
print("unique values in native country")
print(df['native.country'].str.strip().unique())
print("unique values in marital status")
print(df['marital.status'].str.strip().unique())


#task6
print(df.describe())
print(df.describe(include='object'))

#task7
most_common=df['occupation'].value_counts().idxmax()
print("most common occupation is ", most_common)

perc= df['sex'].value_counts(normalize=True) * 100
print("percentage is ", perc)

#task8
df.groupby('education')['education.num'].unique()
print("no inconsistencies")

#task9
print("task 9")
import pandas as pd
import sqlite3

df = pd.read_csv("adult.csv")
conn = sqlite3.connect("adult.db")
df.to_sql("adult_income", conn, if_exists="replace", index=False)

result = pd.read_sql_query(
    "SELECT * FROM adult_income WHERE age > 30", conn
)

print("Original shape:", df.shape)
print("Extracted shape:", result.shape)

conn.close()

#task10
print(df.shape)
#24 exact , 67 without income
#no issues , all valid
```
