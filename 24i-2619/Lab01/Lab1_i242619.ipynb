# Task 9

```python
import pandas as pd
import sqlite3

df_original = pd.read_csv('drive/MyDrive/Colab Notebooks/lab1_DAV_s/adult.csv')

conn = sqlite3.connect('adult_income.db')

df_original.to_sql('adult_income',conn,if_exists='replace',index=False)

query = "SELECT * FROM adult_income WHERE age > 30"


df_sql = pd.read_sql_query(query, conn)

print("SQL Extracted Data:")
print(df_sql.head())
print(' ')
print("nOriginal CSV shape:", df_original.shape)
print("SQL extracted shape:", df_sql.shape)

expected_rows = (df_original['age'] > 30).sum()
print(' ')
print("Expected rows with age > 30:", expected_rows)
print("Actual SQL rows:", df_sql.shape[0])

if df_sql.shape[0] == expected_rows and df_sql.shape[1] == df_original.shape[1]:
    print("SQL extraction is correct")
else:
    print("SQL extraction does not match")

conn.close()
```

# Task 10

### Dataset Overview
The dataset contains information about people's age, education, occupation, workclass, hours worked per week and other personal and employment-related characteristics. The dataset has 48,842 rows and 15 columns after loading the data. It is mainly used to study and predict whether a person's income is less than or equal to 50K or greater than 50K per year.

### Data Quality Issues Found
The main missing values were found in the workclass, occupation, and native-country columns. These columns contained blank entries, while the other columns had little or no missing data. No inconsistent mappings were found between education and education-num and each education level was associated with the same education number. Duplicate records were also checked and handled where necessary.

### Key Observations
The descriptive statistics showed that the average age was around 39 years, while the median age was around 37 years. The average number of hours worked per week was about 40 hours, which was very close to the median of 40 hours. The income classes were imbalanced, with considerably more people earning <=50K than >50K. The frequency analysis also showed that Prof-specialty was one of the most common occupations.

### Cleaning Decisions Made
For the categorical columns workclass, occupation, and native-country, missing values were replaced with "Unknown" rather than deleting the rows. This was done to keep as much of the original data as possible. For any missing numeric values, the median was used because it is less affected by extreme values. The education and education-num columns were checked for consistency, and no corrections were needed.
