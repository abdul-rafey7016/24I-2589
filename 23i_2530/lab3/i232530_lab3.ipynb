```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set global visual style
sns.set_theme(style="whitegrid")

# Load Sample-Superstore dataset (using latin1 encoding for special characters)
# ERROR EXPLANATION: The file 'Sample-Superstore.csv' was not found in the current directory.
# To fix this, please ensure the CSV file is uploaded to your Colab environment
# or provide the correct path to the file.
# For now, a placeholder DataFrame `df` has been created to allow the rest of the code to run.
data = {
    'Order Date': ['2023-01-15', '2023-01-20', '2023-02-10', '2023-02-25', '2023-03-05', '2023-03-12'],
    'Sales': [100, 150, 120, 180, 200, 250],
    'Profit': [10, 15, 12, 18, 20, 25]
}
df = pd.DataFrame(data)

# Parse date and prepare monthly aggregation
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M').astype(str)

# Aggregated monthly sales and profit
monthly_data = df.groupby('YearMonth')[['Sales', 'Profit']].sum().reset_index().sort_values('YearMonth')
```

```python
# 4(a) WCAG Contrast Ratio Function
def get_relative_luminance(rgb_255):
    rgb = [x / 255.0 for x in rgb_255]
    rgb = [c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4 for c in rgb]
    return 0.2126 * rgb[0] + 0.7152 * rgb[1] + 0.0722 * rgb[2]

def get_contrast_ratio(color1_rgb, color2_rgb):
    L1 = get_relative_luminance(color1_rgb)
    L2 = get_relative_luminance(color2_rgb)
    l_max, l_min = max(L1, L2), min(L1, L2)
    return (l_max + 0.05) / (l_min + 0.05)

test_colors = {
    'Diverging Blue (#2b83ba)': (43, 131, 186),
    'Diverging Red (#d7191c)': (215, 25, 28),
    'Grid Gray (#e0e0e0)': (224, 224, 224),
    'Dark Text (#333333)': (51, 51, 51)
}

print("=== WCAG AA Contrast Test against White Background (4.5:1 Target) ===")
for name, rgb in test_colors.items():
    ratio = get_contrast_ratio(rgb, (255, 255, 255))
    status = "PASS" if ratio >= 4.5 else "FAIL"
    print(f"{name}: {ratio:.2f}:1 -> {status}")

# 4(b) Colorblind Accessible Fix with Hatching Patterns
# The 'Sub-Category' column is missing from the current placeholder DataFrame 'df'.
# sub_profit = df.groupby('Sub-Category')['Profit'].sum().reset_index()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Inaccessible Red/Green Chart
# This plot requires 'Sub-Category'. Placeholder added.
axes[0].set_title('Inaccessible: Red vs Green Only (Data Missing)')
axes[0].text(0.5, 0.5, 'Requires "Sub-Category" column in df', horizontalalignment='center', verticalalignment='center', transform=axes[0].transAxes)
# colors_rg = ['green' if p >= 0 else 'red' for p in sub_profit['Profit']]
# axes[0].bar(sub_profit['Sub-Category'], sub_profit['Profit'], color=colors_rg)
# axes[0].tick_params(axis='x', rotation=90)

# Accessible Palette + Pattern Hatching
# This plot requires 'Sub-Category'. Placeholder added.
axes[1].set_title('Accessible: Colorblind Palette + Texture Hatching (Data Missing)')
axes[1].text(0.5, 0.5, 'Requires "Sub-Category" column in df', horizontalalignment='center', verticalalignment='center', transform=axes[1].transAxes)
# bars = axes[1].bar(
#     sub_profit['Sub-Category'], sub_profit['Profit'],
#     color=['#2b83ba' if p >= 0 else '#d7191c' for p in sub_profit['Profit']],
#     edgecolor='black'
# )
# for bar, val in zip(bars, sub_profit['Profit']):
#     if val < 0:
#         bar.set_hatch('//') # Apply texture hatch to negative values

axes[1].axhline(0, color='black', linewidth=1)
# axes[1].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()

# Alt Text Output
# print("\nAlt Text: Bar chart showing profit by sub-category. Positive profits are styled in solid blue bars, while negative profits are styled in red bars with diagonal line hatching textures to ensure visual accessibility without relying on color alone.")
```
