```python
import numpy as np


print("Missing values before replacing ? ")
display(df.isnull().sum())

df.replace('?', np.nan, inplace=True)
print("Unique values in 'workclass' before replacement:")


print('\nMissing values after replacing "?" with NaN:')
display(df.isnull().sum())

print('\nHead of the DataFrame after replacement:')
display(df.head())
```

```python
#Task 2
missing_counts = df.isnull().sum()
missing_percentages = (df.isnull().sum() / len(df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage': missing_percentages
})

missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)

display(missing_info)
```

```python
#Task 3
for col in ['workclass', 'occupation']:
    common_value = df[col].mode()[0]
    df[col].fillna(common_value, inplace=True)

print("Missing values after imputation:")
display(df.isnull().sum())

print("\nHead of the DataFrame after imputation:")
display(df.head())
```

```python
#Task 4
print('Checking for duplicated rows')
print(df.duplicated().sum())
```

```python
#Task 5
for col in ['education','marital.status','native.country']:
    print(f"Unique values for {col}:")
    df[col] = df[col].astype(str).str.strip()
    print(df[col].unique())
    print("\n")
```

```python
#Task 6
print("\nDescriptive Statistics for Numeric Columns:")
display(df.describe().T)

print("\nDescriptive Statistics for Categorical Columns:")
display(df.describe(include='object').T)

print("\nAnalysis of 'age' and 'hours.per.week':")

age_mean = df['age'].mean()
age_median = df['age'].median()
hours_mean = df['hours.per.week'].mean()
hours_median = df['hours.per.week'].median()
print(f"Mean of 'age': {age_mean:.2f}")
print(f"Median of 'age': {age_median}")
print(f"Mean of 'hours.per.week': {hours_mean:.2f}")
print(f"Median of 'hours.per.week': {hours_median}")
```

The mean and median of both columns are approximately equal which mean that there are not many outliers.

```python
print("\n--- Task 8: Cross-Check a Data Quality Assumption ---\n")

education_consistency = df.groupby('education')['education.num'].unique()
print("Consistency check for 'education' and 'education.num':")
print(education_consistency)
inconsistencies = education_consistency[education_consistency.apply(len) > 1]
print(inconsistencies)
print("\nReport on inconsistencies:")
if not inconsistencies.empty:
    print("Inconsistencies found: Each education level should map to a single education.num, but the following show multiple mappings:")
    print(inconsistencies)
else:
    print("No inconsistencies found. Each education level maps consistently to a single education.num.")
```

```python
#Task 9
import sqlite3
import pandas as pd
csv_file = "adult.csv"

original_df = pd.read_csv(csv_file)
print(f"Original CSV Shape: {original_df.shape}")

conn = sqlite3.connect(":memory:")

original_df.to_sql("adult", conn, if_exists="replace", index=False)

sql_query = "SELECT * FROM adult WHERE age > 30"
sql_filtered_df = pd.read_sql_query(sql_query, conn)

csv_filtered_df = original_df[original_df["age"] > 30]

print(f"DataFrame filtered from CSV shape: {csv_filtered_df.shape}")
print(f"DataFrame extracted via SQL shape: {sql_filtered_df.shape}")

if csv_filtered_df.shape == sql_filtered_df.shape:
    print("Verification Successful: Shapes match exactly!")
else:
    print("Verification Failed: Shapes do not match.")

conn.close()
```

## Data Profiling Summary Report

### 1. Dataset Overview

The dataset contains information about individuals to predict income levels. It has **32,561 rows** and **15 columns**, including demographic information (age, sex, marital status, race, native country), educational background (education, education.num), employment details (workclass, occupation, hours.per.week), financial aspects (capital.gain, capital.loss, fnlwgt), and the target variable, 'income'.

### 2. Data Quality Issues Found

*   **Missing Values**: Initially, values represented by '?' were identified. These were replaced with nan values. The columns with missing values and their counts/percentages are:
    *   `occupation`: 1843 missing values (5.66%)
    *   `workclass`: 1836 missing values (5.64%)
    *   `native.country`: 583 missing values (1.79%)

*   **Duplicates**: **24 duplicate rows** were found in the dataset.

*   **Inconsistent Data**: consistency between 'education' and 'education.num'. No inconsistencies were found.

### 3. Key Observations

*   **Income Class Imbalance**: The 'income' distribution is significantly imbalanced:
    *   `<=50K`: 75.92%
    *   `>50K`: 24.08%
    

*   **Most Common Occupation**: The most common occupation in the dataset is **'Prof-specialty'**.

*   **Sex Distribution**: The dataset consists of approximately `66.92% Male` and `33.08% Female` individuals.

*   **Age and Hours per Week**: The mean and median for 'age' (38.58 and 37.0, respectively) and 'hours.per.week' (40.44 and 40.0, respectively) are very close.


### 4. Cleaning Decisions Made

*   **Missing Value Handling**: The '?' character, indicating missing data, was replaced with nan across the DataFrame.

*   **Imputation**: For the 'workclass' and 'occupation' columns, missing values were imputed using the **mode** (most frequent value) of their respective columns. This approach was chosen to preserve the categorical nature of these features.
*   **Whitespace Stripping**: For categorical columns like 'education', 'marital.status', and 'native.country', leading/trailing whitespaces were removed using .str.strip() to ensure consistency and accurate unique values.
