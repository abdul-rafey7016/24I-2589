# Task : 4
##Box Plot: Visual Outlier Detection

```python
sns.boxplot(x='race',y='age',data=df)
plt.xticks(rotation=90)
plt.xlabel('Race')
plt.ylabel('age')
plt.title('Race Vs age')
```

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['age'] < lower) | (df['age'] > upper)]
print(len(outliers))
```

```python
from scipy import stats
df['age'] = stats.zscore(df['age'])
outliers_z = df[df['age'].abs() > 3]
print(len(outliers_z))
```
