```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv('superstore.csv',encoding='cp1252')
df.head(10)
```

```python
#task 1
df['Order Date'] = pd.to_datetime(df['Order Date'])
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(kind='bar', figsize=(12,5))
plt.show()

monthly_sales.plot(kind='line', figsize=(12,6))
plt.show()

monthly_sales.plot(kind='pie', figsize=(12,6))
plt.show()

print("bar chart shows the best visualisation in this because the categories are larger than 7 in number hence the bar chart enables us to look at the data and compare each instace of month in the following year")
```

```python
#task 2
top_5=df.groupby('Sub-Category')['Sales'].sum().nlargest(5)
print(top_5)
top_5.plot(kind='bar',figsize=(12,5))
plt.show()
print("a bar chart is a good choice as it makes comparing the categories more easier while in a pie chart its more difficult to distinguish between categories of similar large values")
```

```python
#revenue
sns.lineplot(x=monthly_sales.index.astype(str), y=monthly_sales.values)
plt.xticks(rotation=45)
plt.xticks(range(0, len(monthly_sales), 3))
plt.show()
print("line graph is better to show the trend over a period of times however a bar graph is useless as it wouldnt show much of a difference between the categories ")
```

```python
#c
sns.histplot(df['Sales'], bins=30)
print("A histogram is the best choice because it shows how the raw order-line sales values are distributed across different ranges and makes it easy to see where most sales values fall. A pie chart would be a weak choice because it is designed for parts of a whole and cannot effectively show the distribution of hundreds or thousands of individual sales values.")
```

```python
sns.scatterplot(x='Discount', y='Profit', data=df)
print("A scatter plot is the best choice because it shows the relationship between the raw discount and profit values, with each point representing an order. A heat map would be a weaker choice because it usually summarizes or aggregates values into groups, which could hide the individual relationship between discount and profit.")
```

```python
#task 3
#a
region_sales = df.groupby('Region')['Sales'].sum().reset_index()
sns.barplot(x='Region', y='Sales', data=region_sales, palette='Set2')

plt.title('Total Sales by Region')
plt.show()
print("A categorical palette gives each region a distinct color because the regions are separate categories, not values along a numerical scale.")
```

```python
region_category = df.groupby(['Region', 'Category'])['Sales'].sum().unstack()
sns.heatmap(region_category, cmap='Blues', annot=True, fmt='.0f')

plt.title('Sales by Region and Category')
plt.show()
print("A sequential palette such as Blues goes from light → dark, allowing you to immediately see which region/category combinations have low vs high sales.")
```

```python
#c
monthly_profit = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Profit'].sum().reset_index()
sns.barplot(x='Order Date', y='Profit', data=monthly_profit, palette='RdBu')
plt.axhline(0, color='black', linewidth=1)
plt.title('Monthly Profit')
plt.xticks(rotation=45)
plt.xticks(range(0, len(monthly_profit), 3))
plt.show()
```

```python

def contrast_ratio(rgb1, rgb2):

    def relative_luminance(rgb):
        rgb = [c / 255 for c in rgb]

        rgb = [
            c / 12.92 if c <= 0.03928
            else ((c + 0.055) / 1.055) ** 2.4
            for c in rgb
        ]

        return (
            0.2126 * rgb[0] +
            0.7152 * rgb[1] +
            0.0722 * rgb[2]
        )

    l1 = relative_luminance(rgb1)
    l2 = relative_luminance(rgb2)

    lighter = max(l1, l2)
    darker = min(l1, l2)

    return (lighter + 0.05) / (darker + 0.05)


# Colors used in Task 3(a)
colors = {
    'green': (102, 194, 165),
    'orange': (252, 141, 98),
    'purple': (141, 160, 203),
    'pink': (231, 138, 195)
}

# White background
background = (255, 255, 255)

# Check each color against white
for name, color in colors.items():

    ratio = contrast_ratio(color, background)

    if ratio >= 4.5:
        print(name, round(ratio, 2), "PASS")
    else:
        print(name, round(ratio, 2), "FAIL")


# Calculate total profit for each sub-category
profit_data = df.groupby('Sub-Category')['Profit'].sum().reset_index()



colors = [
    'red' if x < 0 else 'green'
    for x in profit_data['Profit']
]

plt.figure(figsize=(12, 6))

plt.bar(
    profit_data['Sub-Category'],
    profit_data['Profit'],
    color=colors
)

plt.axhline(0, color='black', linewidth=1)

plt.xlabel('Sub-Category')
plt.ylabel('Profit')
plt.title('Profit by Sub-Category')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()


colors = [
    'red' if x < 0 else 'green'
    for x in profit_data['Profit']
]

hatches = [
    '///' if x < 0 else ''
    for x in profit_data['Profit']
]

plt.figure(figsize=(12, 6))

bars = plt.bar(
    profit_data['Sub-Category'],
    profit_data['Profit'],
    color=colors
)

# Add hatching to loss bars
for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)

plt.axhline(0, color='black', linewidth=1)

plt.xlabel('Sub-Category')
plt.ylabel('Profit')
plt.title('Profit by Sub-Category')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()


# Alt text:
# Bar chart showing profit by sub-category, with green bars representing gains
# and red hatched bars representing losses.
```

```python
# 2. Cherry-picked date window

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values
)

plt.xlim(20, 30)

plt.xticks(rotation=45)

plt.title('Monthly Sales - Cherry-Picked Date Window')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

```python
# 3. Inverted y-axis

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_profit['Order Date'].astype(str),
    monthly_profit['Profit']
)

plt.gca().invert_yaxis()

plt.xticks(rotation=45)
plt.xticks(range(0, len(monthly_profit), 3))

plt.title('Monthly Profit - Inverted Y-Axis')
plt.xlabel('Month')
plt.ylabel('Profit')

plt.show()
```

```python
#task 6
#honest chart
sns.scatterplot(x='Discount', y='Profit', data=df)
plt.axhline(0, color='black')
plt.title('Discount vs Profit')
plt.show()

#misleading
sns.scatterplot(x='Discount', y='Profit', data=df)
plt.ylim(-500, 500)
plt.axhline(0, color='black')
plt.title('Discount vs Profit - Misleading')
plt.show()

print("A scatter plot is appropriate because it shows the relationship between the raw discount and profit values. A neutral color avoids implying extra meaning, while the black zero line distinguishes profit from loss. The misleading version uses a truncated y-axis, which exaggerates the apparent relationship. The chart is accessible because it does not rely on color alone and has clear labels.")
```
