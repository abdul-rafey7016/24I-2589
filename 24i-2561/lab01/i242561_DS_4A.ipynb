```python
import pandas as pd
import numpy as np

df = pd.read_csv("adult.csv")
df.head()
```

## Task 1



```python
print(df.isnull().sum())
```

```python
df.replace(" ?", np.nan, inplace=True)
df.replace("?", np.nan, inplace=True)
print(df.isnull().sum())
```

## Task 2



```python
missing_count=df.isnull().sum()
missing_percent=(df.isnull().sum()/len(df))*100
missing_summary=pd.DataFrame({"Missing Count": missing_count,"Missing %": missing_percent.round(2)}).sort_values("Missing Count", ascending=False)
missing_summary[missing_summary["Missing Count"] > 0]
```

## Task 3


```python
df["workclass"] = df["workclass"].fillna("Unknown")
df["occupation"] = df["occupation"].fillna("Unknown")
df["native.country"] = df["native.country"].fillna("Unknown")
df.isnull().sum()
```

## Task 4



```python
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
cols_no_target = df.columns.drop("income")
print("Duplicates ignoring income:", df.duplicated(subset=cols_no_target).sum())
```

## Task 10

**Dataset Overview:** The dataset is US census records used to predict whether an individual earns more than 50K or less than 50K annually.

**Data Quality Issues Found:**
- Missing values found as "?" in workclass, occuptation and native.country.
- Some exact duplicate rows found and more duplicates found when ignoring the income column.
- whitespace inconsistencies in education, marital.status and native,country.

**Key Observations:**
- Income classes are imbalanced because majority of people earns more than 50k..
- Most common occupation identified was found prof=speciality.
- Mean and median age/hours-per-week are close that means that data is symmetric.

**Cleaning Decisions Made:**
- Replaced "?" with NaN to identify missing values.
- Inserted missing categorical values with "Unknown" instead of dropping rows in order to preserve sample size.
