## Data Profiling Summary Report

### Dataset overview

The dataset contains 32561 rows and 15 columns. It is the Adult Income dataset and contains information about people such as their age, workclass, education, occupation, marital status, relationship, sex, hours worked per week and income. The income column shows whether a person earns <=50K or >50K.

### Data quality issues found

The main data quality issue was missing values represented by the character ?. After replacing ? with NaN, missing values were found in workclass, occupation and native country. Workclass had 1836 missing values, occupation had 1843 and native country had 583. Duplicate rows were also checked using duplicated. Education and education num were checked to make sure that each education level always had the same numerical value and no inconsistencies were found.

### Key observations

The most common occupation was Prof specialty. The mean age was about 38 years while the median was 37 years, so there was only a small difference between them. The mean hours per week was about 41 while the median was 40. The income classes were imbalanced because around 76 percent of the records belonged to the <=50K class while around 24 percent belonged to the >50K class. The dataset therefore contains many more lower income records than higher income records.

### Cleaning decisions made

The ? values were first replaced with proper NaN values so that they could be correctly identified as missing data. Missing values in workclass, occupation and native country were replaced with Unknown. This was done to keep the records instead of removing them and to avoid incorrectly guessing the missing categories. Leading and trailing spaces were removed from categorical columns using strip. Education and education num were also checked and found to be consistent with each other.
