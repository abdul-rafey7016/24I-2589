```python
import seaborn as sns
import matplotlib.pyplot as plt
# TASK 1

df = sns.load_dataset("tips")
df.head()
sns.barplot(data=df, x="day", y="total_bill",color="orange")
plt.show()

```

```python
# Task 2

sns.countplot(data=df, x='day',color="green")
plt.show()
```

```python
# TASK 3
sns.histplot(data=df,x='total_bill',color="purple",kde=True)
mean=df["total_bill"].median()
print(mean)
# it is right skeewed because we see all our values are forming a peak to the right of the graph
```

```python
df.head()
```

```python
# Task 4
sns.boxplot(data=df,x='total_bill',color="pink")
# 7 outliers appear
```

```python
# task 5
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['total_bill'] < lower) | (df['total_bill'] >upper)]
print(len(outliers))

# this is giving 9 but i got 7 outliers visually maybe the dark circle is 3 outliers stacked on eachother
```

```python
# Task 6
from scipy import stats
df['zscore'] = stats.zscore(df['total_bill'])
outliers_z = df[df['zscore'].abs() > 3]
print(len(outliers_z))

# as task 3 shows it is right skewed the reason of difference of outliers in iqr method and zscore is methods differ because 
# they use different reference points (mean vs median) and different measures of spread (standard deviation vs quartiles)
```
