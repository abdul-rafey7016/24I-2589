```python
# Task 1
df = pd.read_csv("adult.csv")
df.info()
df.replace('?',np.nan, inplace=True)
df.isnull().sum()
```

```python
# Task 2
percent_missing = df.isnull().sum() * 100 / len(df)
missing_value_df = pd.DataFrame({'column_name': df.columns,'percent_missing': percent_missing})
descvalue = missing_value_df.sort_values(by='percent_missing', ascending=False)
print(descvalue)
```

```python
# Task 3

# impute 'workclass' with mode as it has highest missing % of data 5.66. Imputing it with mode is better because % is high and we cant remove that much data
df['workclass'].fillna(df['workclass'].mode()[0], inplace=True)

#impute 'occupation' with mode as it has highest missing % of data 5.63. Same as workclass imputation.
df['occupation'].fillna(df['occupation'].mode()[0], inplace=True)

# Impute'native.country' with the mode and it has 1.8% missing values. It keeps data complete without losing meaningful data.
df['native.country'].fillna(df['native.country'].mode()[0], inplace=True)

# check for final values that were null at first
print('\nMissing values after imputation:')
print(df.isnull().sum())
```

```python
#Task 4
exact_duplis = df.duplicated().sum()
print("\nNumber of exact duplicate rows: ", exact_duplis)

# 2. Check for duplicates ignoring the 'income' (target/label) column
ignoreincome = df.drop(columns=['income']).duplicated().sum()
print("Number of duplicate rows ignoring 'income' column: ",ignoreincome)

# Exact duplicates should be removed as exact record matching confirms duplication, however for the target i.e "income" in our case, that can be same for some records but rest of the attributes can be different and hence be of use so shouldnt be discarded.
```

```python
# Task 5

columns_check = ['education', 'marital.status', 'native.country']

# find unique values
print("\nUnique values:")
for col in columns_check:
    print("Column: ",col, "\n", df[col].unique(), "\n")

# remove whitespace now from the columns in list
for col in columns_check:
    df[col] = df[col].str.strip()

# showing unique values after removal of whitespace
print("\nUnique values after stripping whitespace:")
for col in columns_check:
    print("Column: ",col, "\n", df[col].unique(), "\n")
```

```python
# Task 6
print(df.describe())

# with object
print(df.describe(include='object'))

# Report mean and median for 'age' and 'hours.per.week'
print(df['age'].mean())
print(df['age'].median())
print(df['hours.per.week'].mean())
print(df['hours.per.week'].median())


# for age column the mean median are very close showing they are close and symmetrical and no outliers as mean is affected by outliers
# for hours per week the reason is same as mean median differ by 0.44
```

```python
#task 7
# part a
mostcommoccupation = df['occupation'].value_counts().idxmax()
print("\nMost common occupation:", mostcommoccupation)

# part b
sexdist = df['sex'].value_counts(normalize=True) * 100
print("\nPercentage distribution of sex:\n", sexdist)

# part c
incomedist = df['income'].value_counts(normalize=True) * 100
print("\nPercentage distribution of income label:\n", incomedist)


if incomedist['<=50K'] / incomedist['>50K'] > 1.5 or incomedist['>50K'] / incomedist['<=50K'] > 1.5:
    print("The income classes are imbalanced, with higher % of people earning <=50K.")
else:
    print("The income classes appear relatively balanced.")
```

```python
#task 8

# Group by 'education' and find unique edu.num
edumap = df.groupby('education')['education.num'].unique()

# show inconsistencies
inconsistencies = {}
for edu, nums in edumap.items():
    if len(nums) > 1:
        inconsistencies[edu] = nums

if inconsistencies:
    print("\nInconsistencies found between 'education' and 'education.num':")
    for edu, nums in inconsistencies.items():
        print("Education",edu, " maps to multiple 'education.num' values: ",nums)
else:
    print("\nNo inconsistencies found between 'education' and 'education.num'. Each education level maps to a unique number.")
```
